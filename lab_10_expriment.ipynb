10. Implement transformation-based tagging using a set of transformation rules, apply a  simple rule to tag words using python.

```python
# Baseline tagger assigns everything as Noun (NN), then transformation rules correct errors.
def tbl_tagger(sentence):
    words = sentence.split()
    # Baseline: Tag all as NN
    tagged = [[w, 'NN'] for w in words]

    # Transformation Rule: If word ends with 'ly', change NN to RB
    for i in range(len(tagged)):
        if tagged[i][0].endswith('ly'):
            tagged[i][1] = 'RB'
        # Transformation Rule: If word ends with 's' and not first word, could be VBZ/NNS
        elif tagged[i][0].endswith('s') and i > 0:
            tagged[i][1] = 'VBZ'
    return tagged

text = "he runs quickly"
print("TBL Tagged:", tbl_tagger(text))
```
