Write program demonstrates how to perform morphological analysis using the NLTK  library in Python

```python
import nltk
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet', quiet=True)
nltk.download('omw-1.4', quiet=True)

lemmatizer = WordNetLemmatizer()
words = ["running", "cats", "better", "easily", "wolves"]

print(f"{'Word':<10} | {'Lemma (POS: Verb)':<18} | {'Lemma (POS: Adjective)'}")
print("-" * 52)
for word in words:
    lemma_v = lemmatizer.lemmatize(word, pos='v')
    lemma_a = lemmatizer.lemmatize(word, pos='a')
    print(f"{word:<10} | {lemma_v:<18} | {lemma_a}")
```
