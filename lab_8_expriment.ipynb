8. Implement a simple stochastic part-of-speech tagging algorithm using a basic  probabilistic model to assign POS tags using python.

```python
# Simple probabilistic lexical tagging model
# Format: word -> {tag: probability}
stochastic_lexicon = {
    "time": {"NN": 0.7, "VB": 0.3},
    "flies": {"VBZ": 0.8, "NNS": 0.2},
    "like": {"IN": 0.6, "VB": 0.4},
    "an": {"DT": 1.0},
    "arrow": {"NN": 1.0}
}

def stochastic_tag(sentence):
    tagged = []
    for word in sentence.lower().split():
        if word in stochastic_lexicon:
            # Pick tag with maximum probability
            tags_prob = stochastic_lexicon[word]
            best_tag = max(tags_prob, key=tags_prob.get)
            tagged.append((word, best_tag))
        else:
            tagged.append((word, "NN")) # Default noun
    return tagged

sentence = "time flies like an arrow"
print("Stochastic POS Tagging:", stochastic_tag(sentence))
```
