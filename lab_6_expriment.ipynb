Implement a basic N-gram model for text generation. For example, generate text using a  bigram model using python.

```python
import random
from nltk.util import bigrams
from nltk.tokenize import word_tokenize
import nltk
nltk.download('punkt', quiet=True)

corpus = "machine learning is fun and machine learning is powerful. artificial intelligence and machine learning."
tokens = word_tokenize(corpus.lower())

# Build bigram frequency table
bi_list = list(bigrams(tokens))
model = {}
for w1, w2 in bi_list:
    model.setdefault(w1, []).append(w2)

# Generate text
def generate_text(start_word, length=10):
    current = start_word
    result = [current]
    for _ in range(length):
        if current in model:
            next_word = random.choice(model[current])
            result.append(next_word)
            current = next_word
        else:
            break
    return " ".join(result)

print("Generated Text:", generate_text("machine", 8))
```
