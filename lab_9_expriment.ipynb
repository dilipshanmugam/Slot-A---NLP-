9. Implement a rule-based part-of-speech tagging system using regular expressions using  python.

```python
import nltk
from nltk.tokenize import word_tokenize
nltk.download('punkt', quiet=True)

patterns = [
    (r'.*ing$', 'VBG'),               # gerunds
    (r'.*ed$', 'VBD'),                # past tense verbs
    (r'.*es$', 'VBZ'),                # 3rd person singular present
    (r'^-?[0-9]+(.[0-9]+)?$', 'CD'),   # cardinal numbers
    (r'.*able$', 'JJ'),               # adjectives
    (r'.*ly$', 'RB'),                 # adverbs
    (r'.*', 'NN')                     # default nouns
]

regexp_tagger = nltk.RegexpTagger(patterns)
sentence = "running quickly through connected tables"
tokens = word_tokenize(sentence)
print(regexp_tagger.tag(tokens))
```
